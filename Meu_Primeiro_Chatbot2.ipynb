```python
import random # 1. Importamos a biblioteca para sortear as respostas

def chatbot():
    # Listas de respostas (Dicionário de Personalidade)
    respostas_saudacao = [
        "Olá! Sou seu assistente FashionBot. Que bom te ver por aqui!",
        "Oi! Tudo bem? Sou o robô de estilo da Camila. Como posso ajudar?",
        "E aí! Pronto para falar sobre as melhores tendências de moda?"
    ]

    respostas_estilo = [
        "O estilo atual foca muito em conforto e peças sustentáveis.",
        "A tendência 'Quiet Luxury' está muito forte agora, focando em qualidade e minimalismo.",
        "Estamos vendo uma volta forte dos elementos dos anos 90 com um toque moderno.",
        "O estilo 'Streetwear' continua dominando, misturando peças largas com acessórios luxuosos."
    ]

    respostas_tendencias = [
        "Para 2026, as cores vibrantes e tecidos tecnológicos estão em alta!",
        "A grande aposta agora é a moda circular e o upcycling.",
        "Acessórios maximalistas e sapatos plataforma são o destaque desta temporada."
    ]

    # Início da conversa
    print(f"🤖: {random.choice(respostas_saudacao)}") # Sorteia uma saudação
    nome_usuario = input("Você: ")
    print(f"🤖: Prazer em te conhecer, {nome_usuario}! O que vamos explorar hoje?")

    historico = []

    while True:
        mensagem = input(f"{nome_usuario}: ").lower()
        historico.append(mensagem)

        if mensagem == 'sair':
            print(f"🤖: Tchau, {nome_usuario}! Salvei nosso papo na memória. Até mais!")
            break

        elif 'estilo' in mensagem:
            # 2. O random.choice escolhe um item aleatório da nossa lista lá de cima
            print(f"🤖: {random.choice(respostas_estilo)}")

        elif 'tendências' in mensagem or 'tendencia' in mensagem:
            print(f"🤖: {random.choice(respostas_tendencias)}")

        elif 'quem é você' in mensagem:
            print(f"🤖: Eu sou o FashionBot, um projeto de IA desenvolvido pela {nome_usuario} no Google Colab!")

        else:
            print("🤖: Hum, ainda estou treinando meus circuitos para esse assunto. Tente falar sobre 'estilo' ou 'tendências'.")

# Inicia o chatbot
chatbot()
```
