```python
def chatbot():
    print(" Olá! Eu sou o FashionBot. Qual o seu nome?")
    nome_usuario = input("Você: ")

    #  lembra o nome
    print(f"Prazer em te conhecer, {nome_usuario}! Como posso te ajudar hoje?")
    print("(Dica: você pode perguntar sobre 'estilo', 'tendências' ou digitar 'sair')")

    historico = [] # Aqui é onde a "memória" da sessão fica guardada

    while True:
        mensagem = input(f"{nome_usuario}: ").lower()
        historico.append(mensagem) # Adiciona o que você disse à memória

        if mensagem == 'sair':
            print(f" Tchau, {nome_usuario}! Foi ótimo conversar com você.")
            break

        elif 'estilo' in mensagem:
            print(" O estilo atual está focando muito em conforto e peças sustentáveis. Você gosta de algo mais clássico ou moderno?")

        elif 'tendências' in mensagem:
            print(" Para 2026, estamos vendo muitas cores vibrantes e tecidos tecnológicos!")

        elif 'quem é você' in mensagem:
            print(f"Eu sou um protótipo de chatbot criado pela {nome_usuario} para o projeto de IA.")

        else:
            print(" Hum, ainda estou aprendendo sobre isso. Pode tentar falar sobre 'estilo' ou 'tendências'?")

# Para iniciar o robô:
chatbot()
```
