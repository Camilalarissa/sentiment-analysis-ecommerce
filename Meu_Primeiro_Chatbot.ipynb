```python
import random

def salvar_historico(nome, mensagens):
    # Cria um arquivo .txt e escreve o histórico nele
    nome_arquivo = f"historico_{nome}.txt"
    with open(nome_arquivo, "w", encoding="utf-8") as arquivo:
        arquivo.write(f"Histórico de conversa com: {nome}\n")
        arquivo.write("="*30 + "\n")
        for linha in mensagens:
            arquivo.write(linha + "\n")
    print(f"\n[SISTEMA]: Arquivo '{nome_arquivo}' gerado com sucesso!")

def chatbot():
    respostas_saudacao = ["Olá!", "Oi! Tudo bem?", "E aí! Pronto para falar de moda?"]
    respostas_estilo = ["O estilo atual foca em conforto.", "A tendência 'Quiet Luxury' está forte."]

    print(f"🤖: {random.choice(respostas_saudacao)}")
    nome_usuario = input("Você: ")

    # Lista para armazenar o diálogo formatado
    memoria_conversa = []

    while True:
        entrada = input(f"{nome_usuario}: ")
        mensagem = entrada.lower()

        # Guardamos quem falou o quê na nossa lista
        memoria_conversa.append(f"{nome_usuario}: {entrada}")

        if mensagem == 'sair':
            resposta_final = f"🤖: Tchau, {nome_usuario}! Gravando nossa conversa..."
            print(resposta_final)
            memoria_conversa.append(resposta_final)

            # CHAMA A FUNÇÃO DE SALVAR
            salvar_historico(nome_usuario, memoria_conversa)
            break

        elif 'estilo' in mensagem:
            resp = random.choice(respostas_estilo)
            print(f"🤖: {resp}")
            memoria_conversa.append(f"🤖: {resp}")

        else:
            resp = "🤖: Ainda estou aprendendo sobre isso!"
            print(resp)
            memoria_conversa.append(resp)

chatbot()
```
